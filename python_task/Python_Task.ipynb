```python
import pandas as pd
import numpy as np
import plotly.express as px
```

```python
df = pd.read_csv("../data/employees.csv")
df.head()
```

```python
department_count = df["department"].value_counts()

print(department_count)
```

```python
fig = px.bar(
    x=department_count.index,
    y=department_count.values,
    title="Employee Distribution by Department",
    labels={
        "x": "Department",
        "y": "Number of Employees"
    }
)

fig.show()
```

```python
skill_count = df["primary_skill"].value_counts()

print("Skill Distribution:")
print(skill_count)
```

```python
fig = px.bar(
    x=skill_count.index,
    y=skill_count.values,
    title="Employee Skill Distribution",
    labels={
        "x": "Primary Skill",
        "y": "Number of Employees"
    }
)

fig.show()
```

```python
skill_development = df[df["skill_level"] == "Beginner"]

print("Employees Requiring Skill Development:")
display(skill_development)
```

```python
top_performers = df.sort_values(
    by="performance_score",
    ascending=False
).head(5)

print("Top Performing Employees:")
display(top_performers)
```

```python
avg_performance = df.groupby("department")["performance_score"].mean()

print("Average Performance Score by Department:")
print(avg_performance)
```

```python
fig = px.bar(
    x=avg_performance.index,
    y=avg_performance.values,
    title="Average Performance Score by Department",
    labels={
        "x": "Department",
        "y": "Average Performance Score"
    }
)

fig.show()
```

```python
total_employees = len(df)

total_departments = df["department"].nunique()

average_performance = df["performance_score"].mean()

employees_needing_development = len(
    df[df["skill_level"] == "Beginner"]
)

print("========== WORKFORCE INSIGHTS ==========")
print(f"Total Employees: {total_employees}")
print(f"Total Departments: {total_departments}")
print(f"Average Performance Score: {average_performance:.2f}")
print(f"Employees Needing Skill Development: {employees_needing_development}")
print("========================================")
```
